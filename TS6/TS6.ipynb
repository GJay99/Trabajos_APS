# Introducción
Para la realización de esta tarea semanal tuvimos como objetivo resolver distintos filtros pasivos RLC, hallando su transferencia, además de su respuesta en frecuencia en módulo y fase.

Recordemos que un filtro es un circuito el cuál, al ser excitado con una señal, solo deja pasar ciertas componentes frecuenciales de la misma, atenuando o suprimiendo la energía que tenga fuera de su banda de paso.

Los filtros pasivos son aquellos que operan exclusivamente con componentes de esta naturaleza, como resistencias, capacitores e inductores. Este tipo de filtros no requieren de una fuente de alimentación externa, como cuando se utilizan amplificadores operacionales.

Principalmente nos encontraremos con tres tipos de filtros: pasa bajos, pasa altos y pasa banda. Sin embargo, existen otros como los suprime banda. Para definir que tipo de filtro es un circuito, es importante el concepto de frecuencia de corte. Esta es la frecuencia a partir de la cual la potencia normalizada es de -3dB, es decir la mitad de la original. A partir de esto podemos definir que si deja pasar frecuencia menores a su $\omega_c$ será un pasa bajos, si hace lo contrario será un pasa altos, y si tiene dos frecuencias, dejando pasar las componentes intermedias, será un pasa banda.

Por último, consideramos importante definir otros dos términos, $\omega_0$ y Q. La primera hace referencia a la frecuencia natural, dada por los elementos reactivos del sistema, en nuestro caso:
\begin{equation}
\omega_0 = \frac{1}{\sqrt{LC}}
\end{equation}
En un diagrama de polos y ceros, $\omega_0$ nos habla de la norma del vector que va desde el origen de coordenadas hasta el polo, en casos como los nuestros que cuentes con polos complejos conjugados.

Por otro lado el factor de calidad Q nos habla del amortiguamiento y la resonancia de un filtro, y su formula depende del tipo de circuito. Nuevamente en nuestro caso, posee una fuerte relación con el ángulo que se forma entre el $-\sigma$ y el vector del polo en el diagrama.
\begin{equation}
Q = \frac{1}{2.cos(\psi)}
\end{equation}

# Resolución
## Circuito A

</center>

Comenzamos calculando la función transferencia de este circuito, la cuál podemos hallar fácilmente mediante un divisor de tensión de la siguiente forma:
\begin{equation}
T(s) = \frac{V_{out}}{V_{in}} = \frac{Z_2}{Z_1+Z_2}
\end{equation}
Siendo $Z_1$ el total de la impedancia que se encuentra entre un nodo de la $V_{in}$ y uno de la $V_{out}$, mientras que $Z_2$ la impedancia cuyo voltaje tomamos como salida.

En base a esto, resolvemos:
\begin{equation}
T(s)=\frac{\frac{1}{sC}}{\frac{1}{sC}+R+sL}=\frac{\frac{1}{LC}}{\frac{1}{LC}+\frac{sR}{L}+s^2}=\frac{{\omega_0}^2}{{\omega_0}^2+s\frac{\omega_0}{Q}+s^2}
\end{equation}
\begin{equation}
{w_0}^2=\frac{1}{LC} \land \frac{w_0}{Q}=\frac{R}{L} \Longrightarrow Q=L=\frac{1}{\sqrt2} \land 1=\frac{1}{LC} \Longrightarrow C=\sqrt2
\end{equation}
\begin{equation}
T(s)=\frac{1}{1+\sqrt2s+s^2} \land T(jw)=\frac{1}{1+\sqrt2jw-w^2}
\end{equation}

Ahora que ya tenemos la respuesta en frecuencia, podemos calcular su módulo y fase:
\begin{equation}
|T(jw)|=\frac{1}{\sqrt{(1-w^2)^2+(\sqrt2w)^2}}=\frac{1}{\sqrt{1+w^4}}
\end{equation}
\begin{equation}
\measuredangle T(jw)=0-tan^{-1}(\frac{\sqrt2w}{1-w^2})
\end{equation}

Continuamos buscando la frecuencia de corte, para lo cual debemos igualar la transferencia a $\frac{1}{\sqrt{2}}$ y resolver:
\begin{equation}
|T(\omega_c)|=\frac{1}{\sqrt{1+{\omega_c}^4}}=\frac{1}{\sqrt{2}}=>1+{\omega_c}^4=2=>\omega_c= 1 \frac{rad}{s}
\end{equation}
El cuál, por como está configurado este circuito, concuerda exactamente con la frecuencia crítica.

Veamos ahora lo que sucede al graficar el módulo y la fase de la respuesta en frecuencia. Para esto nos será de principal interés lo que sucederá en los siguientes puntos:
\begin{equation}
|T(\omega_0)|=\frac{1}{\sqrt{2}}
\end{equation}
\begin{equation}
|T(0)|=\frac{1}{\sqrt{1+{0}^4}}=1
\end{equation}
\begin{equation}
\lim_{\omega \to \infty}|T(\omega)|=0
\end{equation}
\begin{equation}
\measuredangle T(\omega_0)=-tan^{-1}(\frac{\sqrt2\omega_0}{1-{\omega_0}^2})=-tan^{-1}(\frac{\sqrt2}{0})=-\frac{\pi}{2}
\end{equation}
\begin{equation}
\measuredangle T(0)=-tan^{-1}(\frac{\sqrt2.0}{1-0^2})=-tan^{-1}(0)=0
\end{equation}
\begin{equation}
\lim_{\omega \to \infty}\measuredangle T(\omega)=-\pi
\end{equation}

Lo que vemos es el efecto correspondiente a un pasa bajos de segundo orden, el cuál deja pasar todas las componentes previas a la $\omega_c$ y filtra las mayores. En concecuencia, su ancho de banda será exactamente igual a la frecuencia de corte ya que este estará entre 0 y $1 \frac{rad}{s}$. Además vemos como para frecuencias muy bajas presenta un retardo casi nulo, mientras que a medidas que la misma aumenta asi lo hará el retardo.

## Circuito B
<center>
    <img src="circuito_b.jpeg" width="400" />
</center>
Nuevamente, nos basaremos en el divisor de tensión por impedancia para poder resolver este circuito:
\begin{equation}
T(s)=\frac{sL}{sL+\frac{R}{RsC+1}}=\frac{s^2LRC+sL}{s^2LRC+sL+R}=\frac{s^2+\frac{s}{RC}}{s^2+\frac{s}{RC}+\frac{1}{LC}}=\frac{s^2+s\frac{\omega_0}{Q}}{s^2+s\frac{\omega_0}{Q}+{\omega_0}^2}
\end{equation}

\begin{equation}
{w_0}^2=\frac{1}{LC} \land \frac{w_0}{Q}=\frac{1}{RC} \Longrightarrow C=\frac{1}{\sqrt2} \land L=\sqrt2
\end{equation}
\begin{equation}
T(s)=\frac{s^2+\sqrt2s}{s^2+\sqrt2s+1} \land T(jw)=\frac{-w^2+\sqrt2jw}{-w^2+\sqrt2jw+1}
\end{equation}

Procedemos ahora a ver lo que sucede con el módulo y la fase de esta transferencia:
\begin{equation}
|T(jw)|=\frac{\sqrt{(-w^2)^2+(\sqrt2w)^2}}{\sqrt{(1-w^2)^2+(\sqrt2w)^2}}=\frac{\sqrt{w^4+2.w^2}}{\sqrt{1+w^4}}
\end{equation}
\begin{equation}
\measuredangle T(jw)=tan^{-1}(\frac{\sqrt2w}{-w^2})-tan^{-1}(\frac{\sqrt2w}{1-w^2})=tan^{-1}(\frac{\sqrt2}{-w})-tan^{-1}(\frac{\sqrt2w}{1-w^2})
\end{equation}

Continuamos buscando su frecuencia de corte de igual manera que para el circuito anterior:
\begin{equation}
|T(\omega_c)|=\frac{\sqrt{{\omega_c}^4+2.{\omega_c}^2}}{\sqrt{1+{\omega_c}^4}}=\frac{1}{\sqrt2}=>2.({\omega_c}^4+2.{\omega_c}^2)=1+{\omega_c}^4 => {\omega_c}^4 + 4 . {\omega_c}^2 - 1 = 0
\end{equation}
Hacemos el cambio de variable $x= {\omega_c}^2$:
\begin{equation}
{x}^2 + 4 . x - 1 = 0 => x_1 = \sqrt5 - 2 \land x_2 = -\sqrt5 - 2
\end{equation}
Descartamos la segunda por ser negativa, y recuperamos nuestra variable original:
\begin{equation}
\omega_c = \sqrt{x} = \sqrt{\sqrt5 - 2} \approx 0.486 \frac{rad}{s}
\end{equation}

Graficamos la función transferencia, donde tenemos que tener especial detalle en lo que sucederá en $\omega_0$ ya que:
\begin{equation}
|T(\omega_0)|=\frac{\sqrt{{\omega_0}^4+2.{\omega_0}^2}}{\sqrt{1+{\omega_0}^4}} = \frac{\sqrt{1^4+2.1^2}}{\sqrt{1+1^4}}=\frac{\sqrt3}{\sqrt2}
\end{equation}
Lo cuál es mayor que 1, indicando que la función tendrá un sobre pico y una zona donde amplificará la señal.
\begin{equation}
|T(0)|=\frac{\sqrt{0^4+2.0^2}}{\sqrt{1+0^4}}=\frac{0}{1}=0
\end{equation}
\begin{equation}
\lim_{\omega \to \infty}|T(\omega)|=1
\end{equation}
\begin{equation}
\measuredangle T(\omega_0)=tan^{-1}(\frac{\sqrt2}{-w_0})-tan^{-1}(\frac{\sqrt2w_0}{1-{w_0}^2})= 0.615 rad \approx \frac{\pi}{5}
\end{equation}
\begin{equation}
\measuredangle T(0)=tan^{-1}(\frac{\sqrt2}{0})-tan^{-1}(\frac{\sqrt2.0}{1-{0}^2})=\frac{\pi}{2}
\end{equation}
\begin{equation}
\lim_{\omega \to \infty}\measuredangle T(\omega)= 0
\end{equation}

Justamente, en el módulo, podemos ver el pico esperado en la frecuencia crítica. Además de que vemos el comportamiento de un filtro pasa altos, teniendo una transferencia mayor a -3dB a partir de la frecuencia de corte. Este tipo de filtros tiene un ancho de banda infinito, ya que la frecuencia puede ser tan alta como sea imaginable y, teoricamente, deberia dejarla pasar. También vemos que a medida que aumenta la frecuencia disminuye el retardo, llegando a 0 asintóticamente.

## Circuito C
<center>
    <img src="circuito_c.jpeg" width="400" />
</center>
Para este último circuito también usamos un divisor de tensión, pero de admitancia debido a que facilita el cálculo de una salida con componentes en paralelo. En este caso la formula será:
\begin{equation}
T(s)=\frac{Y_1}{Y_1+Y_2} \land Y_1 = \frac{1}{Z_1}
\end{equation}

Una vez planteada esta ecuación, simplemente reemplazamos:
\begin{equation}
T(s)=\frac{\frac{1}{R}}{\frac{1}{R}+sC+\frac{1}{sL}}=\frac{\frac{sL}{R}}{s^2CL+\frac{sL}{R}+1}=\frac{\frac{s}{RC}}{s^2+\frac{s}{RC}+\frac{1}{CL}}=\frac{s\frac{\omega_0}{Q}}{s^2+s\frac{\omega_0}{Q}+{\omega_0}^2}
\end{equation}

\begin{equation}
{w_0}^2=\frac{1}{CL} \land \frac{w_0}{Q}=\frac{1}{RC} \Longrightarrow C=\frac{1}{\sqrt2} \land L=\sqrt2
\end{equation}

\begin{equation}
T(s)=\frac{\sqrt2s}{s^2+\sqrt2s+1} \land T(jw)=\frac{\sqrt2jw}{-w^2+\sqrt2jw+1}
\end{equation}

\begin{equation}
|T(jw)|=\frac{\sqrt2w}{\sqrt{(\sqrt2w)^2+(1-w^2)^2}}=\frac{\sqrt2w}{\sqrt{1+w^4}}
\end{equation}

\begin{equation}
\measuredangle T(jw)=\frac{\pi}{2}-tan^{-1}(\frac{\sqrt2w}{1-w^2})
\end{equation}

\begin{equation}
|T(\omega_c)|=\frac{\sqrt2.\omega_c}{\sqrt{1+{\omega_c}^4}}=\frac{1}{\sqrt2}=>2.\omega_c = \sqrt{1+{\omega_c}^4} => -{\omega_c}^4 + 4 .{\omega_c}^2 - 1 = 0
\end{equation}
Volvemos a usar el cambio de variable previamente utilizado $x={\omega_c}^2$:
\begin{equation}
-x^2 + 4 x - 1 = 0 => x_1 = 2 - \sqrt{3} \land x_2 = 2 + \sqrt{3}
\end{equation}
En este caso no podemos descartar ninguno, lo cuál el lógico considerando que al ser un filtro pasa banda requiere de una \omega_c inferior y otra superior.
\begin{equation}
\omega_{inf} = \sqrt{x_1} = 0.518 \frac{rad}{seg}
\end{equation}
\begin{equation}
\omega_{sup} = \sqrt{x_2} = 1.932 \frac{rad}{seg}
\end{equation}

En este caso, al tratarse de un pasa banda podemos calcular su ancho de banda de de dos maneras, una es con las frecuencias de corte:
\begin{equation}
BW = \omega_{sup} - \omega_{inf} = 1.932 \frac{rad}{seg} - 0.518 \frac{rad}{seg} = 1.414 \frac{rad}{seg}
\end{equation}
Mientras que también lo podemos obtener como:
\begin{equation}
BW = \frac{\omega_0}{Q} = \frac{1 \frac{rad}{seg}}{\frac{1}{\sqrt2}}  = \sqrt2 \frac{rad}{seg} \approx 1.414 \frac{rad}{seg}
\end{equation}

Por último realizaremos los gráficos, nuevamente prestando atención en lo que sucederá en $\omega_0$, $\omega = 0$ y $\omega \rightarrow \infty$:
\begin{equation}
|T(\omega_0)|=\frac{\sqrt2.\omega_0}{\sqrt{1+{\omega_0}^4}}=\frac{\sqrt2.1}{\sqrt{1+1^4}}=\frac{\sqrt2}{\sqrt{2}}=1
\end{equation}
\begin{equation}
|T(0)|=\frac{\sqrt2.0}{\sqrt{1+0^4}}=\frac{0}{\sqrt{1}}=0
\end{equation}
\begin{equation}
\lim_{\omega \to \infty}|T(\omega)|=\lim_{\omega \to \infty}\frac{\sqrt2.\omega}{\sqrt{1+{\omega}^4}}=0
\end{equation}
\begin{equation}
\measuredangle T(\omega_0)=\frac{\pi}{2}-tan^{-1}(\frac{\sqrt2w_0}{1-{w_0}^2})= 0
\end{equation}
\begin{equation}
\measuredangle T(0)=\frac{\pi}{2}-tan^{-1}(\frac{\sqrt2.0}{1-{0}^2})=\frac{\pi}{2}
\end{equation}
\begin{equation}
\lim_{\omega \to \infty}\measuredangle T(\omega)= -\frac{\pi}{2}
\end{equation}

Lo que vemos es el comportamiento de un filtro pasa banda, donde la transferencia ideal de 1 se encuentra exclusivamente en la frecuencia crítica y posee un ancho de banda reducido, retrasando y adelantando según la frecuencia sea mayor o menor a la $\omega_0$, respectivamente.

# Conclusiones
A lo largo de este trabajo pudimos ver como utilizar los mismos tres elementos en configuraciones distintas puede causar un cambio muy significativo en el comportamiento de un filtro. Incluso, se observo como a pesar de ser componentes pasivos, en el caso B se observó una ganancia. Esto nos muestra que gracias a los comportamientos reactivos es posible realizar una amplificación de amplitud, aunque no así de potencia (algo similar a lo que sucede en un transformador).

Otro detalle interesante que se pudo apreciar es como, al mantener la mismas cantidad de componentes y los mismos factores Q y $\omega_0$, el denominador de todas las transferencias fue exactamente el mismo. Esto trae consigo que tengan exactamente los mismos polos. Pero, al tener distinto numerados, tendrán distintos ceros. Es por ello que los comportamientos difieren.

Por último pudimos ver el efecto que tiene la frecuencia crítica en cada tipo de filtro, notando que no tiene porque ser equivalente a la frecuencia de corte, además de ver los distintos anchos de banda que presentan.
